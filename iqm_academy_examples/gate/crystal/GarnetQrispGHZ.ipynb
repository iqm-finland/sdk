### Writing a Qrisp program

Qrisp provides advanced quantum data types such as QuantumFloat that we will use in this example. In this example we wil encode a 2 it will multiply it with itself and then measure the result. 

Let's run it on a simulator first.

```python
from qrisp import *
a = QuantumFloat(2)
a[:] = 2
b = a*a
b.get_measurement()
```

## Now let's run the same GHZ as we did with qiskit using Qrisp
The logic to write the circuit follows Qiskit.

```python
num_qb = 9
from qrisp import QuantumCircuit

# Create Quantum Registers and Quantum Circuit
qc = QuantumCircuit(num_qb)
qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)
qc.cx(0, 3)
qc.cx(0, 4)
qc.cx(1, 5)
qc.cx(1, 6)
qc.cx(1, 7)
qc.cx(2, 8)

qc.measure(qc.qubits)
print(qc)
```

```python
meas_results = qrisp_garnet.run(qc, shots = 1000)
print(meas_results)
```
